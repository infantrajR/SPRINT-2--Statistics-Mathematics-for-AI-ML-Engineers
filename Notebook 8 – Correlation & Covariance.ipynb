# 1. Correlation
# Definition
Correlation is a statistical measure used to understand the strength and direction of the relationship between two variables.

The correlation value ranges from -1 to +1.

+1 → Perfect positive correlation
0 → No linear correlation
-1 → Perfect negative correlation
Formula
Pearson correlation coefficient:

r = Cov(X, Y) / (σX × σY)

Where:

r = Correlation coefficient
Cov(X, Y) = Covariance between X and Y
σX = Standard deviation of X
σY = Standard deviation of Y
Numerical Example
Suppose:

X = [1, 2, 3, 4, 5]

Y = [2, 4, 6, 8, 10]

When X increases, Y also increases proportionally.

Therefore:

r = +1

This represents a perfect positive correlation.

# Where It Is Used
Correlation is used to:

# Understand relationships between variables
Identify related features
Explore datasets
Support feature selection in AI/ML
# Demonstrate the Concept
If two variables increase together, they have positive correlation.

If one variable increases while the other decreases, they have negative correlation.

If there is no clear linear relationship, the correlation is close to 0.

```python

import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

correlation = np.corrcoef(x, y)[0, 1]

print("Correlation coefficient:", correlation)

plt.scatter(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Positive Correlation")
plt.grid()
plt.show()
```

# Real-World Example
A person's study hours and exam marks may have a positive correlation.

Generally, when study hours increase, exam marks may also increase.

# AI/ML Example
Correlation can be used to understand relationships between features.

For example, in a house price dataset:

House size
Number of rooms
House price
Correlation can help identify which variables are related to the target variable.

# Interpretation
A correlation value close to +1 indicates a strong positive linear relationship.

A value close to -1 indicates a strong negative linear relationship.

A value close to 0 indicates little or no linear relationship.

# Important Point
Correlation does not necessarily mean causation.

Two variables can be correlated without one variable directly causing the other.

# 2. Covariance
# Definition
Covariance is a statistical measure that shows how two variables change together.

It tells us the direction of the relationship between two variables.

Positive covariance → Both variables tend to increase together.
Negative covariance → One variable tends to increase while the other decreases.
Covariance close to 0 → Little or no linear relationship.
Formula
Cov(X, Y) = Σ[(Xi - X̄)(Yi - Ȳ)] / (n - 1)

# Where:

Xi = Individual value of X
Yi = Individual value of Y
X̄ = Mean of X
Ȳ = Mean of Y
n = Number of observations
# Numerical Example
Suppose:

X = [1, 2, 3]

Y = [2, 4, 6]

Mean of X = 2

Mean of Y = 4

Covariance:

= [((1-2)(2-4)) + ((2-2)(4-4)) + ((3-2)(6-4))] / (3-1)

= [2 + 0 + 2] / 2

= 2

So, the covariance is 2, which indicates a positive relationship.

# Where It Is Used
Covariance is used to:

Understand how variables change together

Analyze relationships between variables

Calculate correlation

Understand relationships between features in AI/ML
# Demonstrate the Concept
Covariance mainly tells us the direction of the relationship, while correlation also gives a standardized measure of the strength of the relationship.

```python
import numpy as np

x = np.array([1, 2, 3])
y = np.array([2, 4, 6])

covariance_matrix = np.cov(x, y)

covariance = covariance_matrix[0, 1]

print("Covariance:", covariance)
```

# Real-World Example
A company's advertising expenditure and sales may have positive covariance if higher advertising spending is generally associated with higher sales.

# AI/ML Example
Covariance can be used to understand how two features change together.

For example, in a house price dataset:

House size
House price
If larger houses generally have higher prices, their covariance may be positive.

# Interpretation
Positive covariance means the variables tend to move in the same direction.

Negative covariance means they tend to move in opposite directions.

The magnitude of covariance depends on the units of the variables, so covariance values are not directly standardized.

Correlation is often easier to compare because it always ranges from -1 to +1.

# 3. Positive Correlation
Definition
Positive correlation occurs when two variables tend to move in the same direction.

When one variable increases, the other variable also tends to increase.

Similarly, when one variable decreases, the other variable also tends to decrease.

# Formula
The Pearson correlation coefficient is used to measure positive correlation:

r = Cov(X, Y) / (σX × σY)

For positive correlation:

0 < r ≤ 1

# Numerical Example
Suppose:

X = [1, 2, 3, 4, 5]

Y = [2, 4, 6, 8, 10]

As X increases, Y also increases.

The correlation coefficient is:

r = +1

This represents a perfect positive correlation.

# Where It Is Used
Positive correlation is used to identify variables that tend to increase or decrease together.

# Demonstrate the Concept
A scatter plot of positively correlated variables generally moves upward from left to right.

```python

import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

correlation = np.corrcoef(x, y)[0, 1]

print("Correlation coefficient:", correlation)

plt.scatter(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Positive Correlation")
plt.grid()
plt.show()
```

# Real-World Example
Study hours and exam marks can show positive correlation.

Generally, students who spend more time studying may achieve higher marks.

# AI/ML Example
In a house price dataset, house size and house price may have positive correlation.

As house size increases, house price may also increase.

# Interpretation
A positive correlation means the variables tend to move in the same direction.

A value close to +1 indicates a strong positive linear relationship.

# Important Point
Positive correlation shows an association between variables. It does not prove that one variable causes the other.

# 4. Negative Correlation
# Definition
Negative correlation occurs when two variables tend to move in opposite directions.

When one variable increases, the other variable tends to decrease.

Formula
The Pearson correlation coefficient is used to measure negative correlation:

r = Cov(X, Y) / (σX × σY)

For negative correlation:

-1 ≤ r < 0

# Numerical Example
Suppose:

X = [1, 2, 3, 4, 5]

Y = [10, 8, 6, 4, 2]

As X increases, Y decreases.

The correlation coefficient is:

r = -1

This represents a perfect negative correlation.

# Where It Is Used
Negative correlation is used to identify variables that tend to move in opposite directions.

# Demonstrate the Concept
A scatter plot of negatively correlated variables generally moves downward from left to right.

```python


import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 8, 6, 4, 2])

correlation = np.corrcoef(x, y)[0, 1]

print("Correlation coefficient:", correlation)

plt.scatter(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Negative Correlation")
plt.grid()
plt.show()
```

# Real-World Example
As the price of a product increases, the number of customers purchasing it may decrease.

This can show a negative correlation between price and number of purchases.

# AI/ML Example
In an ML dataset, a feature such as product price may have a negative correlation with the number of units sold.

# Interpretation
A negative correlation means the variables tend to move in opposite directions.

A value close to -1 indicates a strong negative linear relationship.

# Important Point
Negative correlation shows an association between variables. It does not prove that one variable causes the other.

# 5. No Correlation
# Definition
No correlation means there is no clear linear relationship between two variables.

When one variable changes, there is no consistent pattern in how the other variable changes.

Formula
The Pearson correlation coefficient is used to measure linear correlation.

For no linear correlation:

r ≈ 0

# Numerical Example
Suppose:

X = [1, 2, 3, 4, 5]

Y = [3, 1, 5, 2, 4]

There is no clear increasing or decreasing pattern between X and Y.

The correlation coefficient is close to 0.

# Where It Is Used
No correlation is useful for identifying variables that do not have a meaningful linear relationship.

This can help during:

Data analysis
Feature selection
    
Exploratory data analysis
    
# Demonstrate the Concept
When data points are scattered randomly without a clear upward or downward pattern, the variables may have little or no linear correlation.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5])
y = np.array([3, 1, 5, 2, 4])

correlation = np.corrcoef(x, y)[0, 1]

print("Correlation coefficient:", correlation)

plt.scatter(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("No Clear Linear Correlation")
plt.grid()
plt.show()
```

# Real-World Example
A person's shoe size and the number of books they read may have no meaningful linear relationship.

# AI/ML Example
During feature analysis, two features may have a correlation close to zero.

This indicates that there is no clear linear relationship between those two features.

# Interpretation
A correlation close to 0 means there is little or no linear relationship between the variables.

# Important Point
A correlation close to zero means no strong linear relationship. It does not necessarily mean that there is no relationship of any kind. A non-linear relationship can still exist.

# 6. Pearson Correlation
Definition
Pearson correlation measures the strength and direction of the linear relationship between two numerical variables.

The Pearson correlation coefficient is represented by r.

Its value ranges from -1 to +1.

+1 → Perfect positive correlation
0 → No linear correlation
-1 → Perfect negative correlation
Formula
r = Cov(X, Y) / (σX × σY)

Where:

r = Pearson correlation coefficient
Cov(X, Y) = Covariance between X and Y
σX = Standard deviation of X
σY = Standard deviation of Y
Numerical Example
Suppose:

X = [1, 2, 3, 4, 5]

Y = [2, 4, 6, 8, 10]

As X increases, Y also increases proportionally.

Therefore:

r = +1

This represents a perfect positive linear correlation.

# Where It Is Used
Pearson correlation is used when:

Both variables are numerical.
We want to measure a linear relationship.
The data is suitable for measuring linear association.
# Demonstrate the Concept
Pearson correlation gives a standardized value between -1 and +1, making it easy to understand the direction and strength of a linear relationship.

```python

import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

pearson_correlation = np.corrcoef(x, y)[0, 1]

print("Pearson Correlation:", pearson_correlation)

plt.scatter(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Pearson Correlation")
plt.grid()
plt.show()
```

# Real-World Example
A company can analyze the relationship between advertising spending and sales revenue.

Pearson correlation can show whether higher advertising spending is linearly associated with higher sales.

# AI/ML Example
Pearson correlation can be used during exploratory data analysis to identify relationships between numerical features.

For example, we can check the correlation between:

House size and house price
Experience and salary
Study hours and marks
Interpretation
A Pearson correlation close to +1 indicates a strong positive linear relationship.

A value close to -1 indicates a strong negative linear relationship.

A value close to 0 indicates little or no linear relationship.

# Important Point
Pearson correlation measures linear relationships. It does not prove causation.

# 7.Correlation Matrix
# Definition
A correlation matrix is a table that shows the correlation between multiple variables at the same time.

The values range from -1 to +1:

+1 → Strong positive correlation
0 → No linear correlation
-1 → Strong negative correlation
Example
Suppose we have:

Study Hours
Attendance
Marks
The correlation matrix shows how each variable is related to the others.

# Where it is used
To understand relationships between multiple variables.
To identify highly correlated features.
To help in feature selection before building an ML model.

```python
import pandas as pd
import numpy as np

data = {
    "Study_Hours": [2, 4, 6, 8, 10],
    "Attendance": [60, 65, 75, 85, 95],
    "Marks": [40, 50, 65, 80, 90]
}

df = pd.DataFrame(data)

correlation_matrix = df.corr()

print("Correlation Matrix:")
print(correlation_matrix)
```

# Business Example
A company can create a correlation matrix between advertising cost, sales, customers, and revenue to understand how these variables are related.

# Interpretation
Value close to +1 → strong positive relationship.
    
Value close to -1 → strong negative relationship.
    
Value close to 0 → weak or no linear relationship.
# AI/ML Use Case
In machine learning, a correlation matrix can help identify highly related features. This can help understand the dataset and detect redundant features before model training.

Heatmap
Definition
A heatmap is a visual representation of data using different shades of color.

In correlation analysis, a heatmap is used to display the correlation matrix. It makes it easier to identify strong positive and negative relationships between variables.

# Example
If the correlation between Study Hours and Marks is close to +1, the heatmap shows a strong positive relationship.

If the correlation is close to -1, it shows a strong negative relationship.

# Where it is used
To visualize relationships between multiple variables.
To quickly identify highly correlated features.
To understand feature relationships during Exploratory Data Analysis (EDA).

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Study_Hours": [2, 4, 6, 8, 10],
    "Attendance": [60, 65, 75, 85, 95],
    "Marks": [40, 50, 65, 80, 90]
}

df = pd.DataFrame(data)

correlation_matrix = df.corr()

plt.imshow(correlation_matrix, cmap="coolwarm")
plt.colorbar()

plt.xticks(range(len(correlation_matrix.columns)), correlation_matrix.columns)
plt.yticks(range(len(correlation_matrix.columns)), correlation_matrix.columns)

plt.title("Correlation Heatmap")
plt.show()
```

# Business Example
A company can use a heatmap to quickly understand relationships between advertising cost, sales, customers, and revenue.

# Interpretation
Darker positive values → strong positive correlation.
Values near 0 → weak or no linear correlation.
Negative values → negative correlation.
# AI/ML Use Case
In AI/ML, a correlation heatmap is commonly used during Exploratory Data Analysis (EDA) to identify relationships between features and detect highly correlated features before model training.
