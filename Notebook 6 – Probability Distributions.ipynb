# Probability Distributions
# Bernoulli Distribution
# Step 1 – Understand the Concept
# Definition
Bernoulli Distribution is used when a single experiment has only two possible outcomes.

The outcomes are usually represented as:

1 → Success
0 → Failure
For example, a customer either purchases a product or does not purchase it.

Formula
P(X = x) = p^x (1-p)^(1-x)

Where:

p = probability of success
1-p = probability of failure
x = 0 or 1
Where is it used?
Pass or Fail
Yes or No
Purchase or No Purchase
Spam or Not Spam
Machine Works or Fails
Step 2 – Demonstrate the Concept
Numerical Example
Suppose the probability that a customer purchases a product is 0.7.

Probability of Success = 0.7

Probability of Failure = 1 - 0.7 = 0.3

Real-World Example
A customer receives a notification.

There are two possible outcomes:

1 → Customer clicks the notification 0 → Customer does not click the notification

# Business Example
An e-commerce company can record whether a customer purchased a product.

1 → Purchased 0 → Not Purchased

# AI/ML Example
Binary classification can have two classes:

1 → Spam 0 → Not Spam

Step 3 – Implement the Concept

```python
import numpy as np
import matplotlib.pyplot as plt

p = 0.7

outcomes = [0, 1]
probabilities = [1 - p, p]

plt.bar(outcomes, probabilities)

plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.title("Bernoulli Distribution")

plt.xticks([0, 1], ["Failure", "Success"])

plt.show()
```

# Interpretation of Results
The graph shows:

Failure probability = 0.3
Success probability = 0.7
Therefore, success has a higher probability than failure.

# Advantages
Simple to understand.
Useful for binary outcomes.
Easy to implement.
# Limitations
Only two outcomes are possible.
Represents only one trial.

# Interview Questions and Answers
# 1. What is Bernoulli Distribution?
Bernoulli Distribution models one experiment with two possible outcomes.

# 2. How many outcomes are possible?
Two outcomes: success and failure.

# 3. What does p represent?
p represents the probability of success.

# 4. What is the probability of failure?
1 - p.

# 5. Give an AI/ML example.
Spam and Not Spam classification.

# Binomial Distribution
# Step 1 – Understand the Concept
# Definition
Binomial Distribution is used to calculate the probability of getting a specific number of successes in a fixed number of independent trials.

Formula
P(X = k) = C(n,k) × p^k × (1-p)^(n-k)

Where:

n = number of trials
k = number of successes
p = probability of success
Where is it used?
Marketing campaigns
Customer conversions
Quality testing
A/B testing
Repeated success/failure experiments
Step 2 – Demonstrate the Concept
Numerical Example
Suppose:

n = 5 p = 0.6 k = 3

We want the probability of exactly 3 successes.

P(X = 3) = C(5,3) × (0.6)^3 × (0.4)^2

P(X = 3) = 0.3456

Therefore, the probability is 34.56%.

# Real-World Example
A company sends an offer to 5 customers.

Each customer has a 60% chance of purchasing.

We can calculate the probability that exactly 3 customers purchase.

# Business Example
A marketing team can estimate how many customers may respond to a campaign.

# AI/ML Example
It can be used to analyze repeated binary outcomes such as success/failure events.

# Step 3 – Implement the Concept

```python
import numpy as np
import matplotlib.pyplot as plt
from math import comb

n = 5
p = 0.6

k_values = np.arange(0, n + 1)

probabilities = [
    comb(n, k) * (p ** k) * ((1 - p) ** (n - k))
    for k in k_values
]

plt.bar(k_values, probabilities)

plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")

plt.show()
```

# Interpretation of Results
The graph shows the probability of getting 0, 1, 2, 3, 4, or 5 successes from 5 trials.

Advantages
Useful for a fixed number of trials.
Easy to interpret.
Useful for binary experiments.
Limitations
Trials should be independent.
Probability of success should remain constant.
Interview Questions and Answers
# 1. What is Binomial Distribution?
It calculates the probability of a specific number of successes in fixed independent trials.

# 2. What is the difference between Bernoulli and Binomial?
Bernoulli represents one trial, while Binomial represents multiple trials.

# 3. What is n?
n represents the number of trials.

# 4. What is k?
k represents the number of successes.

# 5. What is p?
p represents the probability of success.

# Poisson Distribution
# Step 1 – Understand the Concept
# Definition
Poisson Distribution is used to calculate the probability of a certain number of events occurring within a fixed interval of time or space.

Formula
P(X = k) = (e^-λ × λ^k) / k!

Where:

λ = average number of events
k = number of events
e = Euler's number
Where is it used?
Customer calls per hour
Website requests
Customer arrivals
Machine failures
Number of defects
Step 2 – Demonstrate the Concept
Numerical Example
A customer support center receives an average of 3 calls per hour.

λ = 3

The probability of exactly 2 calls is:

P(X = 2) = (e^-3 × 3^2) / 2!

P(X = 2) ≈ 0.224

Therefore, the probability is approximately 22.4%.

# Real-World Example
A shop receives an average number of customers every hour.

# Poisson Distribution can calculate the probability of receiving a specific number of customers.

# Business Example
A company can estimate the number of customer support calls expected per hour.

# AI/ML Example
Poisson Distribution can be used for modeling count-based events occurring within a fixed interval.

# Step 3 – Implement the Concept

```python
import numpy as np
import matplotlib.pyplot as plt
from math import exp, factorial

lambda_value = 3

k_values = np.arange(0, 11)

probabilities = [
    (exp(-lambda_value) * (lambda_value ** k)) / factorial(k)
    for k in k_values
]

plt.bar(k_values, probabilities)

plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.title("Poisson Distribution")

plt.show()
```

# Interpretation of Results
The graph shows the probability of different numbers of events when the average event rate is 3.

Advantages
Useful for count data.
Useful for event-arrival problems.
Easy to interpret.
Limitations
Assumes events are independent.
Assumes a constant average rate.
Interview Questions and Answers
# 1. What is Poisson Distribution?
It models the number of events occurring in a fixed interval.

# 2. What does λ represent?
λ represents the average number of events per interval.

# 3. What type of data does Poisson Distribution model?
It mainly models count data.

# 4. Give a business example.
Number of customer calls per hour.

# 5. What is the difference between Poisson and Exponential Distribution?
Poisson models the number of events, while Exponential models the waiting time between events.

# Uniform Distribution
# Step 1 – Understand the Concept
# Definition
Uniform Distribution is a probability distribution where values within a specified range have equal probability density.

Formula
For continuous Uniform Distribution:

f(x) = 1 / (b-a)

Where:

a = lower limit
b = upper limit
Where is it used?
Random sampling
Simulation
Random number generation
Games
Testing
Step 2 – Demonstrate the Concept
Numerical Example
Suppose a random value is selected between 0 and 10.

Every interval of the same length has the same probability.

# Real-World Example
Randomly selecting a number between 1 and 100.

# Business Example
A company can randomly select customers from a population for a survey.

# AI/ML Example
Uniform Distribution can be used for random initialization and simulation.

# Step 3 – Implement the Concept

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(0, 10, 1000)

plt.hist(data, bins=20)

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Uniform Distribution")

plt.show()
```

# Interpretation of Results
The values are spread relatively evenly between 0 and 10.

Advantages
Simple to understand.
Useful for random simulation.
Easy to generate using Python.
Limitations
# Real-world data may not follow a uniform pattern.
Assumes equal density across the selected range.
# Interview Questions and Answers
1. What is Uniform Distribution?
It represents a range where equal-length intervals have equal probability.

2. What are a and b?
They represent the lower and upper limits.

3. Where is it used?
It is used in random sampling and simulation.

4. Can Uniform Distribution be discrete?
Yes. It can be discrete or continuous.

5. Give an AI/ML use case.
Random initialization and simulation.

# Normal Distribution
# Step 1 – Understand the Concept
# Definition
Normal Distribution is a bell-shaped and approximately symmetric distribution where most values are concentrated around the mean.

Formula
f(x) = [1 / (σ√(2π))] × e^[-(x-μ)^2 / (2σ^2)]

Where:

μ = mean
σ = standard deviation
Where is it used?
Height measurements
Test scores
Measurement errors
Statistical analysis
AI/ML
Step 2 – Demonstrate the Concept
Numerical Example
Suppose:

Mean = 100 Standard Deviation = 10

According to the Empirical Rule:

68% → within 1 standard deviation 95% → within 2 standard deviations 99.7% → within 3 standard deviations

# Real-World Example
Human height can often be approximately represented by a normal distribution in a sufficiently large population.

# Business Example
A manufacturing company can analyze product measurements around a target value.

# AI/ML Example
Normal Distribution is used in statistical modeling, noise modeling, and probabilistic machine-learning methods.

# Step 3 – Implement the Concept

```python
import numpy as np
import matplotlib.pyplot as plt

mean = 100
std = 10

data = np.random.normal(mean, std, 1000)

plt.hist(data, bins=30)

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Normal Distribution")

plt.show()
```

# Interpretation of Results
The histogram shows an approximately bell-shaped pattern, with most observations concentrated around the mean.

# Advantages
Widely used in statistics.
Easy to analyze.
Useful for modeling many measurement processes.
# Limitations
Not every dataset is normally distributed.
Skewed data may violate the normality assumption.
# Interview Questions and Answers
1. What is Normal Distribution?
It is a bell-shaped and approximately symmetric continuous distribution.

2. What are its main parameters?
Mean and standard deviation.

3. What does mean represent?
The center of the distribution.

4. What does standard deviation represent?
The spread of the distribution.

5. What is the Empirical Rule?
Approximately 68%, 95%, and 99.7% of observations lie within 1, 2, and 3 standard deviations of the mean.

# Self Reflection
I learned how the mean and standard deviation control the center and spread of a normal distribution.

# Exponential Distribution
# Step 1 – Understand the Concept
# Definition
Exponential Distribution is used to model the waiting time until an event occurs.

Formula
f(x) = λe^(-λx), x ≥ 0

Where:

λ = event rate
x = waiting time
Where is it used?
Waiting time
Customer arrivals
Machine failures
Service time
Time between events
Step 2 – Demonstrate the Concept
Numerical Example
Suppose a support center receives an average of 2 calls per hour.

λ = 2

The Exponential Distribution can model the waiting time until the next call.

# Real-World Example
The waiting time until the next customer enters a shop.

# Business Example
A company can analyze the time between customer service requests.

# AI/ML Example
It can be used in time-to-event and reliability modeling.

# Step 3 – Implement the Concept

```python
import numpy as np
import matplotlib.pyplot as plt

lambda_value = 2

data = np.random.exponential(1 / lambda_value, 1000)

plt.hist(data, bins=30)

plt.xlabel("Waiting Time")
plt.ylabel("Frequency")
plt.title("Exponential Distribution")

plt.show()
```

# Interpretation of Results
The histogram is usually right-skewed.

Many observations have short waiting times, while fewer observations have longer waiting times.

# Advantages
Useful for waiting-time problems.
Useful for reliability analysis.
Simple rate-based interpretation.
Limitations
Assumes a constant event rate.
May not fit real data when the event rate changes over time.
# Interview Questions and Answers
1. What is Exponential Distribution?
It models the waiting time until an event occurs.

2. What does λ represent?
λ represents the event rate.

3. What type of variable does it model?
It models continuous waiting time.

4. What is its common shape?
It is usually right-skewed and decreasing.

5. What is the relationship between Poisson and Exponential Distribution?
Poisson models the number of events in an interval, while Exponential models the waiting time between events.
