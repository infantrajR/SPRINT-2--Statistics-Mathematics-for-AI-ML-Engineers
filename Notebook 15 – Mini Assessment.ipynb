# Question 2 – Variance
# Question:
# What is Variance?

# Coding Exercise:
Write a Python program to calculate the variance of the following dataset:

```python
data = [10, 20, 30, 40, 50]

mean = sum(data) / len(data)

squared_differences = []

for value in data:
    squared_differences.append((value - mean) ** 2)

variance = sum(squared_differences) / len(data)

print("Variance:", variance)
```

# Question 4 – Bayes' Theorem
# Question:
Explain Bayes' Theorem with an example.

# Coding Exercise:
Write a Python program to calculate the probability using Bayes' Theorem.

# Example:
A disease affects 1% of people. A test is positive for 90% of people who have the disease and 5% of people who do not have the disease. Calculate the probability that a person actually has the disease when the test result is positive.

```python
p_disease = 0.01
p_positive_given_disease = 0.90
p_positive_given_no_disease = 0.05

p_no_disease = 1 - p_disease

p_positive = (
    p_positive_given_disease * p_disease
    + p_positive_given_no_disease * p_no_disease
)

result = (p_positive_given_disease * p_disease) / p_positive

print("Probability of disease given positive test:", result)
```

# Question 6 – Correlation and Covariance
# Question:
# Explain Correlation and Covariance.

# Coding Exercise:
Write a Python program to calculate the Covariance and Correlation between the following two datasets:

```python
x = [10, 20, 30, 40, 50]
y = [15, 25, 35, 45, 55]

mean_x = sum(x) / len(x)
mean_y = sum(y) / len(y)

covariance = sum((x[i] - mean_x) * (y[i] - mean_y) for i in range(len(x))) / len(x)

std_x = (sum((value - mean_x) ** 2 for value in x) / len(x)) ** 0.5
std_y = (sum((value - mean_y) ** 2 for value in y) / len(y)) ** 0.5

correlation = covariance / (std_x * std_y)

print("Covariance:", covariance)
print("Correlation:", correlation)
```

# Question 7 – Skewness
# Question:
# What is Skewness?

# Coding Exercise:
Write a Python program to calculate the skewness of the following dataset:

```python
data = [10, 20, 20, 30, 40, 100]

mean = sum(data) / len(data)

standard_deviation = (
    sum((value - mean) ** 2 for value in data) / len(data)
) ** 0.5

skewness = (
    sum((value - mean) ** 3 for value in data) / len(data)
) / (standard_deviation ** 3)

print("Skewness:", skewness)
```

# Question 8 – IQR
# Question:
# Explain IQR (Interquartile Range).

# Coding Exercise:
Write a Python program to calculate the IQR of the following dataset:

```python
data = [10, 20, 30, 40, 50, 60, 70, 80]

data.sort()

n = len(data)
mid = n // 2

lower_half = data[:mid]
upper_half = data[mid:]

q1 = (lower_half[1] + lower_half[2]) / 2
q3 = (upper_half[1] + upper_half[2]) / 2

iqr = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
```

# Question 9 – p-value
# Question:
# What is a p-value?

# Coding Exercise:
Write a Python program to calculate the p-value for a one-sample t-test using the following sample data:

```python
data = [52, 48, 51, 49, 50, 53, 47, 51, 49, 50]

hypothesized_mean = 50
```

```python
import math

data = [52, 48, 51, 49, 50, 53, 47, 51, 49, 50]
hypothesized_mean = 50

n = len(data)
mean = sum(data) / n

variance = sum((x - mean) ** 2 for x in data) / (n - 1)
std_dev = math.sqrt(variance)

t_value = (mean - hypothesized_mean) / (std_dev / math.sqrt(n))

print("Mean:", mean)
print("T-value:", t_value)
```

# Question 12 – Detect Outliers Using IQR Method
# Question:
# Write Python code to detect outliers in a dataset using the IQR (Interquartile Range) method.

```python
data = [10, 12, 15, 18, 20, 22, 25, 30, 100]

data.sort()

n = len(data)
mid = n // 2

lower_half = data[:mid]
upper_half = data[mid + 1:]

q1 = lower_half[len(lower_half) // 2]
q3 = upper_half[len(upper_half) // 2]

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = []

for value in data:
    if value < lower_bound or value > upper_bound:
        outliers.append(value)

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Outliers:", outliers)
```

# Question 13 – Calculate Correlation Between Two Variables
# Question:
# Write Python code to calculate the correlation between two variables.

```python
x = [10, 20, 30, 40, 50]
y = [15, 25, 35, 45, 55]

mean_x = sum(x) / len(x)
mean_y = sum(y) / len(y)

numerator = sum(
    (x[i] - mean_x) * (y[i] - mean_y)
    for i in range(len(x))
)

denominator_x = sum((value - mean_x) ** 2 for value in x) ** 0.5
denominator_y = sum((value - mean_y) ** 2 for value in y) ** 0.5

correlation = numerator / (denominator_x * denominator_y)

print("Correlation:", correlation)
```
