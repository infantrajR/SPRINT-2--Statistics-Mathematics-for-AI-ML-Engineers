# Statistical Data Visualization

# 1. Histogram
Purpose
A histogram is used to show the distribution of numerical data by dividing values into intervals called bins.

It helps us understand how frequently values occur and what the overall shape of the data looks like.

# When to Use
When analyzing numerical data.
    
When we want to understand the distribution of values
    
To identify patterns such as skewness, spread, and possible outliers.
# Advantages
Easy to understand.
    
Shows the distribution of numerical data clearly.
    
Helps identify the center and spread of data.
    
Can show skewness and unusual values.
# Limitations
The appearance depends on the number of bins.
    
It is mainly suitable for numerical data.
    
It does not show the exact value of every observation.
# Python Example
We can use Matplotlib's plt.hist() to create a histogram.

```python
import matplotlib.pyplot as plt

exam_scores = [45, 50, 52, 55, 60, 62, 65, 67, 70, 72, 75, 78, 80, 82, 85, 88, 90, 92]

plt.hist(exam_scores, bins=5)

plt.xlabel("Exam Score")
plt.ylabel("Frequency")
plt.title("Distribution of Exam Scores")

plt.show()
```

# Interpretation
The histogram shows how the exam scores are distributed across different score ranges.

Each bar represents a range of exam scores, and the height of the bar represents how many observations fall within that range.

If most values are concentrated around one range, that range has a higher frequency.

# Business Example
A company can use a histogram to analyze the distribution of customer purchase amounts.

It can help identify whether most customers spend small, medium, or large amounts.

# AI/ML Use Case
Histograms are useful during Exploratory Data Analysis (EDA) to understand feature distributions before preprocessing and model training.

# 2. Bar Chart
Purpose
A bar chart is used to compare values across different categories.

Each category is represented by a bar, and the length or height of the bar represents its value.

# When to Use
When comparing different categories.
    
When the data is categorical.
    
To compare quantities such as sales, customers, or product counts.
# Advantages
Simple and easy to understand.
    
Makes category comparisons clear
    
Works well with a small or moderate number of categories.
# Limitations
Too many categories can make the chart difficult to read.
    
It is not ideal for showing continuous distributions.
    
Very small differences may be difficult to notice.
# Python Example
We can use Matplotlib's plt.bar() to create a bar chart.

```python
import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Headphones"]
sales = [120, 180, 90, 150]

plt.bar(products, sales)

plt.xlabel("Product")
plt.ylabel("Number of Sales")
plt.title("Product Sales")

plt.show()
```

# Interpretation
The bar chart compares the number of sales for different products.

The height of each bar represents the sales quantity.

For example, if the Mobile bar is the highest, it means Mobile has the highest number of sales among the products shown.

# Business Example
A company can use a bar chart to compare monthly sales across different products or branches.

# AI/ML Use Case
Bar charts are useful during Exploratory Data Analysis (EDA) to understand categorical features, class distributions, and target categories before building a model.

# 3. Line Chart
Purpose
A line chart is used to show changes or trends over a continuous sequence, especially over time.

It connects data points with lines so we can easily see whether values are increasing, decreasing, or staying stable.

# When to Use
When analyzing trends over time.
    
For time-series data.
    
When comparing changes between different time periods.
    
# Advantages
Clearly shows trends and patterns.
    
Easy to identify increases and decreases.
    
Useful for time-series data.
# Limitations
Too many lines can make the chart difficult to understand.
    
It is not ideal for unrelated categorical data.
    
Connecting points can sometimes suggest a continuous relationship where one may not exist.
# Python Example
We can use Matplotlib's plt.plot() to create a line chart.

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [100, 120, 115, 140, 160, 180]

plt.plot(months, sales, marker="o")

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales Trend")

plt.show()
```

# Interpretation
The line chart shows how sales change from January to June.

An upward line indicates that sales are increasing.

A downward line indicates that sales are decreasing.

In this example, the overall trend shows that monthly sales are increasing.

# Business Example
A company can use a line chart to monitor monthly revenue and identify growth or decline over time.

# AI/ML Use Case
Line charts are useful for visualizing time-series data such as stock values, sales, website traffic, temperature, or model performance over training iterations.

# 4. Scatter Plot
Purpose
A scatter plot is used to show the relationship between two numerical variables.

Each point represents one observation with an x-value and a y-value.

# When to Use
When studying the relationship between two numerical variables.
    
To identify positive or negative relationships.
    
To identify clusters and possible outliers.
# Advantages
Shows relationships clearly.
    
Helps identify patterns and trends.
    
Makes unusual observations easy to notice.
    
Useful for checking correlation visually.
# Limitations
It can become difficult to read when there are too many points.
    
It mainly shows relationships between two variables.
    
Correlation shown in a scatter plot does not prove causation.
# Python Example
We can use Matplotlib's plt.scatter() to create a scatter plot.

```python
import matplotlib.pyplot as plt

hours_studied = [1, 2, 3, 4, 5, 6, 7, 8]
exam_scores = [45, 50, 55, 60, 65, 72, 78, 85]

plt.scatter(hours_studied, exam_scores)

plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.title("Hours Studied vs Exam Score")

plt.show()
```

# Interpretation
The scatter plot shows the relationship between hours studied and exam scores.

In this example, as study hours increase, exam scores generally increase.

This indicates a positive relationship between the two variables.

However, a relationship does not automatically mean that one variable causes the other.

# Business Example
A company can use a scatter plot to study the relationship between advertising expenditure and sales.

# AI/ML Use Case
Scatter plots are useful during Exploratory Data Analysis (EDA) to identify relationships between features, check possible correlations, detect outliers, and understand patterns before model training.

# 5. Pie Chart
Purpose
A pie chart is used to show how different categories contribute to a whole.

Each slice represents a category, and the size of the slice represents its proportion of the total.

# When to Use
When showing percentages or proportions.
    
When comparing a small number of categories.
    
When all categories together represent a complete whole.
# Advantages
Easy to understand for a small number of categories.
    
Clearly shows proportions
    
Useful for simple percentage comparisons.
# Limitations
Difficult to compare similar-sized slices.
    
Too many categories make it confusing.
    
Not suitable for showing trends over time.
Bar charts are often easier for precise comparisons.
# Python Example
We can use Matplotlib's plt.pie() to create a pie chart.

```python
import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Headphones"]
sales = [120, 180, 90, 150]

plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%"
)

plt.title("Product Sales Distribution")

plt.show()
```

# Interpretation
The pie chart shows the percentage contribution of each product to the total sales.

A larger slice represents a larger share of total sales.

For example, if Mobile has the largest slice, it contributes the highest percentage of sales among the products.

# Business Example
A company can use a pie chart to show the percentage of total sales contributed by different product categories.

# AI/ML Use Case
Pie charts can be useful for understanding class or category proportions during Exploratory Data Analysis.

For example, we can visualize the percentage of customers in different customer segments.

# 6. Density Plot
Purpose
A density plot shows the distribution of numerical data using a smooth curve.

It is similar to a histogram, but instead of bars, it uses a smooth line to represent the distribution.

# When to Use
When studying the distribution of numerical data.
    
To understand the shape of the data.
    
To compare distributions of different groups.
    
To identify peaks, spread, and skewness.
# Advantages
Shows the distribution smoothly.
    
Makes it easier to compare multiple distributions.
    
Helps identify peaks and overall data shape.
# Limitations
The result depends on the smoothing method and bandwidth
    
It may be harder for beginners to understand than a histogram.
    
It can hide individual observations.
# Python Example
We can use Pandas plot(kind="density") to create a density plot.

```python
import pandas as pd
import matplotlib.pyplot as plt

exam_scores = [45, 50, 52, 55, 60, 62, 65, 67, 70, 72, 75, 78, 80, 82, 85, 88, 90, 92]

df = pd.DataFrame({
    "Exam_Score": exam_scores
})

df["Exam_Score"].plot(kind="density")

plt.xlabel("Exam Score")
plt.title("Density Plot of Exam Scores")

plt.show()
```

# Interpretation
The density curve shows where the exam scores are more concentrated.

A higher part of the curve means that more observations are concentrated around that range.

The shape of the curve can help us understand whether the data is approximately symmetric or skewed.

# Business Example
A company can use a density plot to understand the distribution of customer spending and identify where most customers' spending values are concentrated.

# AI/ML Use Case
Density plots are useful during Exploratory Data Analysis to understand feature distributions and compare the distributions of different groups before model training.

# 7. Box Plot
Purpose
A box plot summarizes the distribution of numerical data using the minimum, Q1, median, Q3, and maximum values.

It also helps identify possible outliers.

# When to Use
When comparing distributions of different groups.
    
To identify outliers.
    
To understand the spread of data.
    
To compare median values between groups.
# Advantages
Quickly shows the center and spread of data.
    
Makes outliers easy to identify.
    
Useful for comparing multiple groups
    
# Limitations
Does not show the complete distribution shape
    
Individual data values are not visible.

Some details may be hidden compared with a histogram.
# Python Example
We can use Matplotlib's boxplot() function to create a box plot.

```python

import matplotlib.pyplot as plt

store_a = [120, 130, 125, 140, 135, 150, 128, 132, 145, 180]
store_b = [100, 110, 105, 115, 120, 108, 112, 118, 125, 130]

plt.boxplot([store_a, store_b])

plt.xticks([1, 2], ["Store A", "Store B"])
plt.ylabel("Sales")
plt.title("Sales Distribution by Store")

plt.show()
```

# Interpretation
The line inside the box represents the median.

The box represents the middle 50% of the data, from Q1 to Q3.

Points outside the normal whisker range can indicate possible outliers.

In this example, Store A has a higher spread and contains a high sales value that may be an outlier.

# Business Example
A company can use box plots to compare sales distributions across different stores and identify unusual sales values.

# AI/ML Use Case
Box plots are useful during Exploratory Data Analysis to detect outliers and understand feature spread before preprocessing and model training.

# 8. Violin Plot
Purpose
A violin plot shows the distribution and shape of numerical data for one or more groups.

It combines ideas from a box plot and density plot.

# When to Use
When comparing distributions between groups.
    
To understand the shape of the data
    
To identify where values are concentrated.
    
To identify possible multiple peaks in the data.
# Advantages
Shows the distribution shape clearly.
    
Useful for comparing multiple groups.
    
Can show more information than a simple box plot.
# Limitations
Can be harder to understand than a box plot.
    
The shape depends on the smoothing method.
    
Small datasets may give misleading distribution shapes.
# Python Example
We can use Matplotlib's violinplot() function.

```python

import matplotlib.pyplot as plt

store_a = [120, 130, 125, 140, 135, 150, 128, 132, 145, 180]
store_b = [100, 110, 105, 115, 120, 108, 112, 118, 125, 130]

plt.violinplot(
    [store_a, store_b],
    showmeans=True,
    showmedians=True
)

plt.xticks([1, 2], ["Store A", "Store B"])
plt.ylabel("Sales")
plt.title("Sales Distribution by Store")

plt.show()
```

# Interpretation
The width of the violin shows where the data is more concentrated.

Wider areas mean more values are concentrated around that range.

The plot also shows the median and mean, helping us understand the center of the data.

In this example, Store A and Store B have different sales distributions.

# Business Example
A company can use violin plots to compare customer spending distributions across different regions or stores.

# AI/ML Use Case
Violin plots can be used during Exploratory Data Analysis to compare feature distributions between different classes or groups before building a machine learning model.

# 9. Pair Plot
Purpose
A pair plot shows the relationships between multiple numerical variables in one visualization.

It creates scatter plots between pairs of variables and shows the distribution of each variable on the diagonal.

# When to Use
During Exploratory Data Analysis (EDA).
    
To understand relationships between multiple features.
    
To identify patterns, trends, and outliers.
    
To check whether features may be correlated.
# Advantages
Shows many feature relationships in one visualization.
    
Helps identify patterns and outliers quickly.
    
Useful for understanding a dataset before machine learning.
# Limitations
Can become crowded when there are many features.
    
Mainly useful for numerical variables.
    
Large datasets can make the visualization difficult to read.
# Python Example
We can create a pair plot using Pandas scatter_matrix().

```python
import pandas as pd
import matplotlib.pyplot as plt
from pandas.plotting import scatter_matrix

data = {
    "Hours_Studied": [2, 4, 5, 7, 8],
    "Attendance": [60, 65, 70, 85, 90],
    "Exam_Score": [50, 55, 65, 80, 90]
}

df = pd.DataFrame(data)

scatter_matrix(df, figsize=(8, 8))

plt.suptitle("Pair Plot of Student Features")
plt.show()
```

# Interpretation
The pair plot shows the relationship between every pair of numerical features.

For example, we can observe the relationship between Hours_Studied and Exam_Score.

The diagonal shows the distribution of each individual feature.

# Business Example
A company can use a pair plot to study relationships between sales, advertising spending, customer visits, and revenue.

# AI/ML Use Case
Pair plots are useful during EDA to identify relationships, correlations, clusters, and possible outliers before selecting features for a machine learning model.

# 10. Heatmap
Purpose
A heatmap represents numerical values using different colors.

It is commonly used to understand patterns in a matrix, such as a correlation matrix.

# When to Use
To visualize a correlation matrix.
    
To identify strong and weak relationships.
    
To compare values in a matrix
    
To quickly identify patterns in data.
# Advantages
Easy to identify patterns using colors.
    
Can display many values in a compact form.
    
Useful for understanding correlations.
# Limitations
Too many variables can make the heatmap difficult to read.
    
Color differences may be difficult to interpret.
    
A correlation shown in a heatmap does not mean causation.
# Python Example
We can use Matplotlib to create a heatmap from a correlation matrix.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Hours_Studied": [2, 4, 5, 7, 8],
    "Attendance": [60, 65, 70, 85, 90],
    "Exam_Score": [50, 55, 65, 80, 90],
    "Sleep_Hours": [8, 7, 7, 6, 6]
}

df = pd.DataFrame(data)

correlation_matrix = df.corr()

plt.imshow(correlation_matrix, cmap="coolwarm")

plt.colorbar()

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

plt.title("Correlation Heatmap")
plt.show()
```

# Interpretation
The heatmap uses colors to represent the strength and direction of relationships.

Values closer to +1 show strong positive correlation.

Values closer to -1 show strong negative correlation.

Values close to 0 show weak or no linear correlation.

In this example, we can use the heatmap to quickly understand the relationships between study hours, attendance, exam score, and sleep hours.

# Business Example
A company can use a heatmap to identify relationships between sales, advertising cost, customer visits, and revenue.

# AI/ML Use Case
Heatmaps are commonly used during Exploratory Data Analysis to understand feature relationships and identify highly correlated features before machine learning model training.
