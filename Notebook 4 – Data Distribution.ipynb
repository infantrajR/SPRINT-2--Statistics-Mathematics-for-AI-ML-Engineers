# Data Distribution

# Normal Distribution
# Step 1 – Understand the Concept
# Definition
Normal distribution is a continuous probability distribution in which most values are concentrated around the mean.

It has a bell-shaped curve and is symmetric around the mean.

In a normal distribution:

Mean = Median = Mode
Most observations are near the center.
Fewer observations occur as we move away from the center.
Formula
The probability density function of a normal distribution is:

f(x) = (1 / (σ√(2π))) × e^(-(x-μ)² / (2σ²))

Where:

μ = Mean
σ = Standard deviation
x = Observation
π = Mathematical constant
Step 2 – Demonstrate the Concept
Numerical Example
Suppose the heights of students in a large group approximately follow a normal distribution.

Mean = 165 cm Standard deviation = 5 cm

Most students will have heights close to 165 cm.

Very short and very tall students will occur less frequently.

# Real-World Example
Human height is often approximately normally distributed within a suitable population and age group.

Most people are close to the average height, while extremely short or extremely tall values are less common.

# Business Example
A manufacturing company produces components with a target weight.

If the production process is stable, the component weights may approximately follow a normal distribution around the target value.

The company can use this distribution to monitor production consistency.

# AI/ML Example
Many statistical and machine-learning techniques use the idea of normally distributed data.

Normal distribution is also important when analyzing model errors and performing statistical inference.

# How does this demonstrate the concept?
The observations are concentrated around the mean and become less frequent as they move away from the center, creating a bell-shaped curve.

# Step 3 – Implement the Concept

```python
%pip install matplotlib
import numpy as np
import matplotlib.pyplot as plt

mean = 165
std = 5

data = np.random.normal(mean, std, 1000)

plt.hist(data, bins=30)
plt.title("Normal Distribution")
plt.xlabel("Height")
plt.ylabel("Frequency")
plt.show()
```

# What Does the Program Do?
The program generates 1,000 random values using a normal distribution with a mean of 165 and a standard deviation of 5.

It then displays the values using a histogram.

#c What Statistical Concept Does It Demonstrate?
It demonstrates the bell-shaped pattern of a normal distribution.

# How Do We Interpret the Output?
Most values should appear close to 165.

Values farther away from 165 occur less frequently.

The histogram should approximately form a bell-shaped pattern.

# Where Is It Used in AI/ML?
Normal distributions are used in statistical modeling, data analysis, probability calculations, and understanding model errors.

# Uniform Distribution
# Step 1 – Understand the Concept
# Definition
Uniform distribution is a distribution where all values within a specified range have equal probability.

In a continuous uniform distribution, every value within the interval has the same probability density.

It generally produces a flat-shaped distribution.

# Formula
For a continuous uniform distribution between a and b:

f(x) = 1 / (b - a)

for a ≤ x ≤ b.

# Step 2 – Demonstrate the Concept
Numerical Example
Suppose a random number is selected between 1 and 10.

In a uniform distribution, values within this range have equal probability density.

# Real-World Example
A simple example is generating a random number between two limits using a random-number generator designed to produce a uniform distribution.

# Business Example
A company may use uniform random sampling when selecting records randomly from a defined population, so that each record has an equal chance of selection.

# AI/ML Example
Uniform distributions can be used when generating random values for simulations or initializing certain model parameters.

# How does this demonstrate the concept?
The generated values are spread relatively evenly across the selected range rather than concentrating around a central value.

# Step 3 – Implement the Concept    

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(1, 10, 1000)

plt.hist(data, bins=20)
plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

# What Does the Program Do?
The program generates 1,000 random values between 1 and 10 using a uniform distribution.

It displays the values using a histogram.

# What Statistical Concept Does It Demonstrate?
It demonstrates uniform distribution.

# How Do We Interpret the Output?
The histogram should show values distributed relatively evenly across the range from 1 to 10.

# Where Is It Used in AI/ML?
Uniform distributions can be used in simulations, random sampling, and some parameter initialization methods.

# Gaussian Distribution
# Step 1 – Understand the Concept
# Definition
Gaussian distribution is another name commonly used for the normal distribution.

It is a continuous probability distribution with a bell-shaped and symmetric curve.

The distribution is described mainly by:

Mean (μ)
Standard deviation (σ)
The mean determines the center of the distribution.

The standard deviation determines how widely the values are spread.

Formula
f(x) = (1 / (σ√(2π))) × e^(-(x-μ)² / (2σ²))

Step 2 – Demonstrate the Concept
Numerical Example
Suppose:

Mean = 50 Standard deviation = 10

Most observations will be concentrated around 50.

Values around 40 and 60 are also common, while values far from 50 are less frequent.

# Real-World Example
Measurement errors in some controlled processes can often be approximately modeled using a Gaussian distribution.

# Business Example
A company can analyze production measurements to determine whether a process is approximately centered around its target value.

# AI/ML Example
Gaussian distributions are used in statistical models and machine-learning techniques such as Gaussian Naive Bayes and Gaussian mixture models.

How does this demonstrate the concept?
The data is centered around the mean and becomes less frequent as the distance from the mean increases.

# Step 3 – Implement the Concept

```python
import numpy as np
import matplotlib.pyplot as plt

mean = 50
std = 10

data = np.random.normal(mean, std, 1000)

plt.hist(data, bins=30)
plt.title("Gaussian Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

# What Does the Program Do?
The program generates 1,000 random values using a Gaussian distribution.

# What Statistical Concept Does It Demonstrate?
It demonstrates the bell-shaped Gaussian distribution.

# How Do We Interpret the Output?
Most values should be near the mean of 50.

The frequency decreases as values move farther away from the mean.

# Where Is It Used in AI/ML?
Gaussian distributions are used in probabilistic models, Gaussian Naive Bayes, Gaussian mixture models, and statistical analysis.

# Left Skew
# Step 1 – Understand the Concept
# Definition
A left-skewed distribution is a distribution where the longer tail extends toward the left side.

It is also called a negatively skewed distribution.

In many left-skewed datasets:

Mean < Median

The majority of observations are concentrated toward the higher-value side, while a smaller number of low values create the longer left tail.

Key Point
Left Skew = Longer tail on the left side.

# Step 2 – Demonstrate the Concept
Numerical Example
Consider exam scores:

40, 70, 75, 80, 82, 85, 88, 90, 92

Most scores are high, but a few low scores create a longer tail toward the left.

# Real-World Example
If most students perform very well in an easy exam but a few students score very low, the score distribution can become left-skewed.

# Business Example
Suppose most products receive high customer ratings, but a few products receive very low ratings.

Those low ratings can create a left tail.

# AI/ML Example
Skewness is useful during EDA because highly skewed features may need transformations before being used in some statistical or ML models.

# How does this demonstrate the concept?
Most observations are concentrated toward the higher values, while the smaller values extend the distribution toward the left.

# Step 3 – Implement the Concept

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.beta(5, 2, 1000)

plt.hist(data, bins=30)
plt.title("Left-Skewed Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

# What Does the Program Do?
The program generates values using a beta distribution with parameters that produce a left-skewed shape.

# What Statistical Concept Does It Demonstrate?
It demonstrates a distribution with a longer tail toward the left.

# How Do We Interpret the Output?
Most values are concentrated toward the higher side, while fewer lower values form the left tail.

# Where Is It Used in AI/ML?
Skewness analysis is used during EDA to understand feature distributions and decide whether transformations may be useful.

# Right Skew
# Step 1 – Understand the Concept
# Definition
A right-skewed distribution is a distribution where the longer tail extends toward the right side.

It is also called a positively skewed distribution.

In many right-skewed datasets:

Mean > Median

Most observations are concentrated toward the lower-value side, while a smaller number of high values create the longer right tail.

Key Point
Right Skew = Longer tail on the right side.

# Step 2 – Demonstrate the Concept
Numerical Example
Consider customer spending:

100, 120, 150, 180, 200, 250, 300, 1000

Most customers spend smaller amounts, but a few customers spend much more.

The high values create a long right tail.

# Real-World Example
Personal income is often right-skewed because many people have moderate incomes while a smaller number of people have very high incomes.

# Business Example
An e-commerce company may have many customers with small purchases and a few customers with very large purchases.

This can create a right-skewed purchase-amount distribution.

# AI/ML Example
Right-skewed features are common in real datasets.

During EDA, analysts may use transformations such as log transformation when appropriate to reduce strong skewness.

# How does this demonstrate the concept?
Most observations are on the lower side, while a few large observations extend the tail toward the right.

# Step 3 – Implement the Concept

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.exponential(scale=2, size=1000)

plt.hist(data, bins=30)
plt.title("Right-Skewed Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

# What Statistical Concept Does It Demonstrate?
It demonstrates positive or right skewness.

# How Do We Interpret the Output?
Most observations are concentrated toward the lower values.

A smaller number of large values create a longer tail toward the right.

# Where Is It Used in AI/ML?
Skewness is checked during EDA, and suitable transformations may be applied to highly skewed numerical features.

# Symmetric Distribution
# Step 1 – Understand the Concept
# Definition
A distribution is symmetric when its left and right sides have approximately the same shape around the center.

If we draw a vertical line through the center, both sides look like mirror images.

For a perfectly symmetric normal distribution:

Mean = Median = Mode

Key Point
Symmetric Distribution = Left side and right side have similar shapes.

# Step 2 – Demonstrate the Concept
Numerical Example
Consider:

10, 20, 30, 40, 50, 60, 70

The values are distributed evenly around the center value of 40.

# Real-World Example
A measurement process that produces values approximately equally distributed around a target can show a symmetric distribution.

# Business Example
If a company's employee performance scores are distributed similarly on both sides of the average score, the distribution is approximately symmetric.

# AI/ML Example
Symmetry is useful during EDA because it helps us understand the shape of a feature distribution and whether skewness is present.

# How does this demonstrate the concept?
The observations are balanced around the center instead of having a longer tail on one side.

# Kurtosis
# Step 1 – Understand the Concept
# Definition
Kurtosis is a statistical measure that describes the shape of the tails of a distribution and how much extreme-value behavior the distribution has compared with a reference distribution.

In practical data analysis, kurtosis is often used to identify whether a dataset has relatively heavier or lighter tails.

There are commonly three descriptions:

Mesokurtic – Similar tail behavior to the normal distribution.
Leptokurtic – Relatively heavier tails and more extreme observations.
Platykurtic – Relatively lighter tails and fewer extreme observations.
Important Note
Different software and textbooks may use different conventions for kurtosis, especially whether they report kurtosis or excess kurtosis.

In pandas, the kurtosis() method reports excess kurtosis. For a normal distribution, excess kurtosis is approximately 0.

# Step 2 – Demonstrate the Concept
Numerical Example
Suppose two datasets have similar means and standard deviations.

Dataset A has relatively few extreme observations.

Dataset B has more extreme observations.

Dataset B can have higher kurtosis because its distribution has heavier tails.

# Real-World Example
Consider daily stock returns.

Most daily returns may be close to zero, but occasionally there may be very large positive or negative movements.

Such extreme observations can contribute to higher kurtosis.

# Business Example
A financial company can analyze the kurtosis of returns to understand the presence of extreme movements in financial data.

# AI/ML Example
Kurtosis can be used during EDA to identify features that may contain extreme observations or heavy tails.

# How does this demonstrate the concept?
Kurtosis focuses on the tail behavior of a distribution and helps us understand the presence of extreme observations.

# Step 3 – Implement the Concept

```python
import numpy as np
from scipy.stats import kurtosis

data = np.array([10, 12, 13, 14, 15, 16, 17, 18, 20, 50])

excess_kurtosis = kurtosis(data)

print("Excess Kurtosis:", excess_kurtosis)
```

# What Does the Program Do?
The program calculates the excess kurtosis of the dataset using SciPy.

# What Statistical Concept Does It Demonstrate?
It demonstrates kurtosis as a measure related to the tail behavior of a distribution.

# How Do We Interpret the Output?
A positive excess kurtosis indicates relatively heavier tails compared with the normal reference distribution.

The value should be interpreted together with the shape of the data and other descriptive statistics.

# Where Is It Used in AI/ML?
Kurtosis can be used during EDA to identify heavy-tailed features and investigate possible extreme observations.

# Empirical Rule (68–95–99.7)
# Step 1 – Understand the Concept
# Definition
The Empirical Rule describes how data is distributed around the mean when the data approximately follows a normal distribution.

# Approximately:

68% of observations lie within 1 standard deviation of the mean.
95% of observations lie within 2 standard deviations of the mean.
99.7% of observations lie within 3 standard deviations of the mean.
Formula / Rule
Within 1 standard deviation:

μ ± 1σ → approximately 68%

Within 2 standard deviations:

μ ± 2σ → approximately 95%

Within 3 standard deviations:

μ ± 3σ → approximately 99.7%

# Important Note
The Empirical Rule applies to approximately normal or bell-shaped distributions. It should not automatically be applied to strongly skewed distributions.

# Step 2 – Demonstrate the Concept
Numerical Example
Suppose exam scores approximately follow a normal distribution.

Mean = 70 Standard deviation = 10

Within 1 standard deviation:

70 ± 10

Range = 60 to 80

Approximately 68% of students are expected to have scores between 60 and 80.

Within 2 standard deviations:

70 ± 20

Range = 50 to 90

Approximately 95% of students are expected to have scores between 50 and 90.

Within 3 standard deviations:

70 ± 30

Range = 40 to 100

Approximately 99.7% of students are expected to have scores between 40 and 100.

# Real-World Example
Suppose the fill volume of bottles produced by a machine is approximately normally distributed.

The Empirical Rule can help estimate how many bottles are expected to fall within one, two, or three standard deviations of the target average.

# Business Example
A manufacturing company can use the rule to understand the expected range of production measurements and identify unusually extreme observations.

# AI/ML Example
The Empirical Rule can help analysts understand normally distributed features and identify observations that are unusually far from the mean.

How does this demonstrate the concept?
The rule connects the mean and standard deviation to expected percentages of observations in a normal distribution.

# Step 3 – Implement the Concept

```python
import numpy as np

mean = 70
std = 10

range_1 = (mean - std, mean + std)
range_2 = (mean - 2 * std, mean + 2 * std)
range_3 = (mean - 3 * std, mean + 3 * std)

print("Within 1 Standard Deviation:", range_1, "-> Approximately 68%")
print("Within 2 Standard Deviations:", range_2, "-> Approximately 95%")
print("Within 3 Standard Deviations:", range_3, "-> Approximately 99.7%")
```

# What Does the Program Do?
The program calculates the ranges corresponding to one, two, and three standard deviations from the mean.

# What Statistical Concept Does It Demonstrate?
It demonstrates the Empirical Rule (68–95–99.7).

# How Do We Interpret the Output?
For a mean of 70 and standard deviation of 10:

Approximately 68% of observations are between 60 and 80.
Approximately 95% are between 50 and 90.
Approximately 99.7% are between 40 and 100.
# Where Is It Used in AI/ML?
The Empirical Rule can help analyze approximately normal features, understand unusual observations, and interpret standard deviation ranges during EDA.
