# Statistics in Machine Learning

# 1. Data Cleaning
Purpose
Data cleaning is the process of finding and correcting problems in a dataset before using it for machine learning.

Statistics helps us understand the data and identify problems such as missing values, incorrect values, duplicates, unusual values, and inconsistent data.

How Statistics is Used
We can use statistical measures such as:

Mean

Median

Mode

Minimum and maximum

Standard deviation

Percentiles

These help us understand whether the data contains unusual or incorrect values.

# When to Use
Data cleaning is used before training a machine learning model.

# Advantages
Improves data quality.
    
Helps reduce errors.
    
Can improve model performance.
    
Makes the dataset more reliable.
# Limitations
Cleaning can take time.
    
Incorrect cleaning decisions can remove useful information.
    
Statistics alone cannot identify every type of data error.
# Python Example
We can use statistical summary methods to understand the data.

```python
import pandas as pd

data = {
    "Age": [22, 25, 28, 30, 35],
    "Salary": [25000, 30000, 35000, 40000, 45000]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

print("\nStatistical Summary:")
print(df.describe())
```

# Interpretation
The describe() function gives statistical information such as:

Count
Mean

Standard deviation

Minimum

25th percentile

Median

75th 

Maximum
These values help us understand the dataset and identify possible data quality problems.

# Business Example
A company can analyze employee age and salary data to identify incorrect or unusual records before using the data for analysis.

# AI/ML Use Case
In machine learning, clean and reliable data is important because poor-quality data can negatively affect model training and predictions.

# 2. Missing Value Handling
Purpose
Missing value handling is the process of identifying and treating missing data in a dataset.

Statistics helps us decide how to replace missing values using measures such as mean, median, and mode.

# How Statistics is Used
Mean can be used for numerical data when there are no strong outliers.
    
Median can be used when the data is skewed or contains outliers.
    
Mode can be used for categorical data.
When to Use
Missing value handling is used before training a machine learning model because many ML algorithms cannot directly work with missing values.

# Advantages
Keeps useful records instead of removing them.
    
Helps create a complete dataset.
    
Can improve model training.
# Limitations
Replacing values can change the original data distribution
    
Mean imputation can be affected by outliers.
    
A wrong method can introduce bias.
# Python Example
We can use the median to fill a missing numerical value.

```python
import pandas as pd

data = {
    "Age": [22, 25, None, 30, 35],
    "Salary": [25000, 30000, 35000, None, 45000]
}

df = pd.DataFrame(data)

print("Before Handling Missing Values:")
print(df)

df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

print("\nAfter Handling Missing Values:")
print(df)
```

# Interpretation
The missing values in Age and Salary are replaced using their respective median values.

The median is useful because it is less affected by extreme values compared with the mean.

# Business Example
A company may have missing salary information for some employees. The median salary can be used to fill the missing values when appropriate.

# AI/ML Use Case
Missing value handling is an important preprocessing step because incomplete data can cause errors or reduce the quality of machine learning predictions.

# 3. Outlier Detection
Purpose
Outlier detection is the process of identifying values that are very different from the other observations in a dataset.

Statistics provides methods such as IQR and Z-score to detect unusual values.

# How Statistics is Used
Two common statistical methods are:

IQR Method: identifies values outside the lower and upper bounds.
    
Z-Score Method: measures how far a value is from the mean in terms of standard deviations.
# When to Use
Outlier detection is used during data preprocessing and Exploratory Data Analysis.

# Advantages
Helps identify unusual observations.
    
Can improve data quality.
    
Helps prevent extreme values from affecting some models.
# Limitations
Not every outlier is an error.
Removing valid outliers can lose useful information.
    
Different methods may identify different outliers.
# Python Example
We can use the Z-score method to identify values that are far from the mean.

```python
import numpy as np

sales = np.array([100, 110, 105, 120, 115, 108, 112, 500])

mean = np.mean(sales)
std = np.std(sales)

z_scores = (sales - mean) / std

print("Sales:", sales)
print("Z-scores:", z_scores)

outliers = sales[np.abs(z_scores) > 2]

print("Possible Outliers:", outliers)
```

# Interpretation
The Z-score tells us how far each sales value is from the mean.

A value with a large absolute Z-score can be considered a possible outlier.

In this example, 500 is much higher than the other sales values, so it may be identified as a possible outlier.

However, we should check the reason for the value before removing it.

# Business Example
A company may identify an unusually high sales transaction. It could be a data entry error or a genuine large customer order.

# AI/ML Use Case
Outlier detection is useful during preprocessing because extreme values can affect statistical models and some machine learning algorithms.

# 4. Feature Engineering
Purpose
Feature engineering is the process of creating new features or transforming existing features to make the data more useful for a machine learning model.

Statistics helps us understand the data and decide which transformations may be useful.

How Statistics is Used
We can use statistical information such as:

Mean and median to understand numerical features.
    
Standard deviation to understand data spread.
    
Correlation to identify relationships between features.
    
Frequency and proportions for categorical features.
# When to Use
Feature engineering is used during data preprocessing when the original features are not sufficient or need transformation.

# Advantages
Can improve model performance.
    
Helps models understand useful patterns.
    
Can make complex information easier for a model to use.
# Limitations
Poorly designed features can reduce model performance.
    
Creating too many features can increase complexity.
    
Statistical relationships do not always represent meaningful real-world relationships.
# Python Example
We can create a new feature from existing features.

For example, we can calculate the price per square foot from house price and house area.

```python
import pandas as pd

data = {
    "House_Area": [1000, 1200, 1500, 1800, 2000],
    "House_Price": [200000, 240000, 300000, 360000, 400000]
}

df = pd.DataFrame(data)

df["Price_Per_Sqft"] = df["House_Price"] / df["House_Area"]

print(df)
```

# Interpretation
A new feature called Price_Per_Sqft is created from House_Price and House_Area.

This new feature gives us additional information that may be useful for predicting house prices.

# Business Example
A real estate company can create features such as price per square foot, property age, or number of rooms from existing property data.

# AI/ML Use Case
Feature engineering helps machine learning models use meaningful information from raw data and can improve the model's ability to identify patterns.

# 5. Feature Scaling
Purpose
Feature scaling is the process of bringing numerical features to a similar scale.

Statistics such as mean and standard deviation are used in some scaling methods.

Two common methods are:

Standardization: transforms data using mean and standard deviation.
    
Min-Max Scaling: transforms values to a fixed range, usually 0 to 1.
    
# How Statistics is Used
Standardization uses the mean and standard deviation of a feature.

The standardized value shows how far a value is from the mean in terms of standard deviations.

# When to Use
Feature scaling is especially useful for algorithms that depend on distance or feature magnitude, such as:

K-Nearest Neighbors
    
K-Means
    
Support Vector Machines
    
Gradient-based models
    
# Advantages
Puts features on comparable scales.
    
Prevents large-scale features from dominating distance-based algorithms.
    
Can improve optimization for some models.
# Limitations
Not necessary for every algorithm.


Scaling does not remove outliers.
    
The scaling parameters should be calculated using training data only.
# Python Example
We can use StandardScaler to standardize numerical features.

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

data = {
    "Age": [20, 25, 30, 35, 40],
    "Salary": [20000, 30000, 40000, 50000, 60000]
}

df = pd.DataFrame(data)

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)

scaled_df = pd.DataFrame(
    scaled_data,
    columns=df.columns
)

print("Original Data:")
print(df)

print("\nScaled Data:")
print(scaled_df)
```

# Interpretation
After standardization, the features are transformed to a similar scale.

The transformed values represent how far each observation is from the feature's mean in terms of standard deviations.

For example, a positive value means the observation is above the mean, while a negative value means it is below the mean.

# Business Example
Suppose a company uses customer age and annual income to group customers. Since income has much larger numerical values than age, scaling can prevent income from dominating distance-based calculations.

# AI/ML Use Case
Feature scaling is commonly used before algorithms such as KNN, K-Means, SVM, and neural networks when differences in feature scales can affect the model

# 6. Data Normalization
Purpose
Data normalization is the process of transforming numerical values to a common range or distribution.

A common method is Min-Max Normalization, which usually converts values to a range between 0 and 1.

How Statistics is Used
Min-Max normalization uses the minimum and maximum values of a feature.

# Formula:

Normalized Value = (X - Minimum) / (Maximum - Minimum)

# When to Use
Normalization is useful when:

Features have different ranges.
    
The machine learning algorithm is sensitive to feature scale.
    
We want values within a fixed range.
# Advantages
Converts features to a common range.
    
Useful for distance-based algorithms.
    
Makes numerical features easier to compare.
# Limitations
Sensitive to outliers.
    
New values outside the original minimum and maximum can affect the scaling.
    
Not required for every machine learning algorithm.
# Python Example
We can use MinMaxScaler from Scikit-learn.

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

data = {
    "Age": [20, 25, 30, 35, 40],
    "Salary": [20000, 30000, 40000, 50000, 60000]
}

df = pd.DataFrame(data)

scaler = MinMaxScaler()

normalized_data = scaler.fit_transform(df)

normalized_df = pd.DataFrame(
    normalized_data,
    columns=df.columns
)

print("Original Data:")
print(df)

print("\nNormalized Data:")
print(normalized_df)
```

# Interpretation
The values of both features are transformed into the range 0 to 1.

The minimum value becomes 0 and the maximum value becomes 1.

The other values are converted proportionally between these limits.

# Business Example
A company may have customer age ranging from 18–60 and income ranging from ₹20,000–₹2,00,000. Normalization can bring both features to a similar range.

# AI/ML Use Case
Normalization can be useful for machine learning algorithms such as KNN, K-Means, neural networks, and other algorithms where feature scale affects the result.

# 7. Model Evaluation
Purpose
Model evaluation is the process of checking how well a machine learning model performs on data.

Statistics provides different measures to compare the model's predictions with the actual values.

How Statistics is Used
For regression, common statistical measures include:

Mean Absolute Error (MAE)
    
Mean Squared Error (MSE)

Root Mean Squared Error (RMSE)

R² Score

For classification, common measures include:

Accuracy
    
Precision

Recall

F1-Score

When to Use
Model evaluation is used after training a machine learning model to understand its performance on unseen data.

# Advantages
Helps measure model performance.
    
Helps compare different models.
    
Helps identify prediction errors.
    
# Limitations
One metric may not give the complete picture.
    
The appropriate metric depends on the problem.
    
Evaluation can be misleading if the test data is not representative.
# Python Example
We can use MAE and R² to evaluate a regression model.

```python
from sklearn.metrics import mean_absolute_error, r2_score

actual = [100, 200, 300, 400, 500]
predicted = [110, 190, 310, 380, 510]

mae = mean_absolute_error(actual, predicted)
r2 = r2_score(actual, predicted)

print("Mean Absolute Error:", mae)
print("R² Score:", r2)
```

# Interpretation
MAE shows the average absolute difference between actual and predicted values.

A lower MAE generally means the predictions are closer to the actual values.

R² Score indicates how much of the variation in the target variable is explained by the model.

A higher R² generally indicates a better fit for the evaluated data, but it should not be used alone.

# Business Example
A company can evaluate a sales prediction model by comparing predicted sales with actual sales.

# AI/ML Use Case
Model evaluation helps an AI/ML engineer select and improve models by measuring their performance on unseen test data.

# 8. Linear Regression
Purpose
Linear Regression is a supervised machine learning algorithm used to predict a continuous numerical value.

Statistics is used to understand the relationship between independent variables and the target variable.

The model tries to find the best-fitting line between the input and output.

# How Statistics is Used
Linear Regression uses statistical concepts such as:

Mean
    
Variance

Correlation

Least Squares

R²
The model finds coefficients that minimize the difference between actual and predicted values.

# When to Use
When the target is numerical.
    
When there is a reasonably linear relationship between features and target
    
For predicting values such as price, sales, or temperature.
# Advantages
Simple and easy to interpret.
    
Fast to train.
    
Shows the relationship between features and target.
# Limitations
Assumes a linear relationship
    
Sensitive to outliers.
    
May perform poorly when the relationship is strongly non-linear.
# Python Example
We can use LinearRegression from Scikit-learn to predict house prices.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

data = {
    "Area": [1000, 1200, 1500, 1800, 2000],
    "Price": [200000, 240000, 300000, 360000, 400000]
}

df = pd.DataFrame(data)

X = df[["Area"]]
y = df["Price"]

model = LinearRegression()
model.fit(X, y)

predicted_price = model.predict([[1600]])

print("Predicted Price for 1600 sq.ft:", predicted_price[0])
print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)
```

# Interpretation
The model learns the relationship between house area and house price.

The coefficient shows how much the predicted price changes when the house area increases by one unit.

The model can then use the learned relationship to predict the price of a new house.

# Business Example
A real estate company can use house area, number of rooms, and location-related features to predict house prices.

# AI/ML Use Case
Linear Regression is commonly used for numerical prediction problems such as sales forecasting, price prediction, demand estimation, and revenue prediction.

# 9. Logistic Regression
Purpose
Logistic Regression is a supervised machine learning algorithm mainly used for classification problems.

It predicts the probability that an observation belongs to a particular class.

For binary classification, the output is usually between 0 and 1.

How Statistics is Used
Logistic Regression uses statistical concepts such as:

Probability
Odds

Log-odds

Maximum likelihood estimation

Classification threshold

The model converts the predicted probability into a class using a threshold such as 0.5.

# When to Use
When the target is categorical.
    
For binary classification such as Yes/No or 0/1.
    
When we want probability-based predictions.
# Advantages
Simple and easy to interpret.
    
Provides probabilities
    
Works well as a baseline classification model.
# Limitations
Works best when the relationship between features and log-odds is reasonably linear.
    
Can be affected by multicollinearity and outliers.
    
May not perform well for complex non-linear patterns without suitable feature engineering.
# Python Example
We can use Logistic Regression to predict whether a customer will churn.

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression

data = {
    "Support_Calls": [1, 2, 3, 4, 5, 6, 7, 8],
    "Churn": [0, 0, 0, 0, 1, 1, 1, 1]
}

df = pd.DataFrame(data)

X = df[["Support_Calls"]]
y = df["Churn"]

model = LogisticRegression()
model.fit(X, y)

new_customer = [[5]]

probability = model.predict_proba(new_customer)[0][1]
prediction = model.predict(new_customer)[0]

print("Churn Probability:", probability)
print("Predicted Class:", prediction)
```


# Interpretation
The model predicts the probability that the customer will churn.

The predict_proba() function gives the probability for each class.

The predict() function converts the probability into a class based on the model's classification threshold.

In this example:

0 represents No Churn.
1 represents Churn.
# Business Example
A telecom company can use customer information such as support calls, contract length, and monthly charges to predict whether a customer may leave the service.

# AI/ML Use Case
Logistic Regression is commonly used for classification tasks such as customer churn prediction, fraud detection, spam detection, and medical classification.

# 10. Naive Bayes
Purpose
Naive Bayes is a supervised machine learning algorithm mainly used for classification problems.

It is based on Bayes' Theorem and calculates the probability of a class using the available features.

It assumes that the features are conditionally independent given the class.

How Statistics is Used
Naive Bayes uses statistical concepts such as:

Probability

Conditional probability

Bayes' Theorem

Prior 

Likelihood

Posterior probability

# When to Use
For classification problems.
    
When working with text or categorical features.
    
For applications such as spam detection and text classification.
# Advantages
Simple and fast.
    
Works well with high-dimensional data.
    
Requires relatively little training data.
    
Very useful for text classification.
# Limitations
The feature independence assumption may not always be true.
    
Performance can decrease when features are strongly dependent.
    
Zero-frequency values can cause probability problems, which can be handled using smoothing.
# Python Example
We can use GaussianNB when the numerical features are assumed to follow a Gaussian distribution.

```python
import pandas as pd
from sklearn.naive_bayes import GaussianNB

data = {
    "Study_Hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "Attendance": [60, 65, 70, 75, 80, 85, 90, 95],
    "Pass": [0, 0, 0, 1, 1, 1, 1, 1]
}

df = pd.DataFrame(data)

X = df[["Study_Hours", "Attendance"]]
y = df["Pass"]

model = GaussianNB()
model.fit(X, y)

new_student = [[5, 82]]

prediction = model.predict(new_student)
probability = model.predict_proba(new_student)

print("Predicted Class:", prediction[0])
print("Class Probabilities:", probability[0])
```

# Interpretation
The Naive Bayes model calculates the probability of each class using the input features.

In this example, the model uses Study_Hours and Attendance to predict whether a student will pass.

The class with the higher predicted probability is selected as the prediction.

# Business Example
An email service can use word frequencies and other features to calculate the probability that an email is spam or not spam.

# AI/ML Use Case
Naive Bayes is widely used for text classification tasks such as spam detection, sentiment classification, document classification, and news categorization.

# 11. Clustering
Purpose
Clustering is an unsupervised machine learning technique that groups similar data points together.

Statistics helps us understand the similarity, distance, and distribution of data points.

A common clustering algorithm is K-Means Clustering.

How Statistics is Used
Clustering uses concepts such as:

Mean

Distance

Variance

Data distribution

Within-cluster variation

In K-Means, the mean of each cluster is used as its centroid.

# When to Use
When there is no target variable.
    
When we want to find natural groups in data.
    
For customer segmentation.
    
For grouping similar observations.
    
Advantages
Does not require labeled data.
    
Helps discover hidden groups.
    
Useful for customer segmentation and exploratory analysis
    
# Limitations
K-Means requires the number of clusters to be selected.
    
Sensitive to outliers.
    
Results can change depending on initial centroids
    
Works best when clusters are reasonably separated.
# Python Example
We can use K-Means to group customers based on their annual income and spending score.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

data = {
    "Annual_Income": [20, 22, 25, 70, 72, 75, 40, 42, 45],
    "Spending_Score": [80, 85, 78, 20, 25, 18, 50, 55, 48]
}

df = pd.DataFrame(data)

X = df[["Annual_Income", "Spending_Score"]]

model = KMeans(n_clusters=3, random_state=42, n_init=10)
df["Cluster"] = model.fit_predict(X)

print(df)

plt.scatter(
    df["Annual_Income"],
    df["Spending_Score"],
    c=df["Cluster"]
)

plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("Customer Clusters")

plt.show()
```

# Interpretation
K-Means groups customers based on their similarity in annual income and spending score.

Each customer is assigned to one cluster.

The cluster centers represent the average position of the customers within each group.

# Business Example
A retail company can group customers into segments such as high-income/high-spending, high-income/low-spending, and medium-income customers.

# AI/ML Use Case
Clustering is used for customer segmentation, recommendation systems, anomaly detection, image segmentation, and exploratory data analysis.

# 12. Principal Component Analysis (PCA)
Purpose
Principal Component Analysis (PCA) is a statistical technique used for dimensionality reduction.

It transforms many correlated features into a smaller number of new features called principal components while trying to preserve as much variation in the data as possible.

How Statistics is Used
PCA uses statistical concepts such as:

Mean
    
Variance
    
Covariance
    
Correlation
    
Eigenvalues and eigenvectors
    
The principal components are created in directions that capture the largest amount of variance in the data.

# When to Use
When a dataset has many numerical features.
    
When features are correlated.
    
To reduce the number of features.
    
To visualize high-dimensional data.
# Advantages
Reduces the number of features.
    
Can reduce computational complexity.
    
Helps visualize high-dimensional data.
    
Can reduce redundancy from correlated features.
# Limitations
Principal components can be difficult to interpret.
    
Important information may be lost when reducing dimensions.
    
Features should generally be scaled before PCA when they are measured on different scales.
# Python Example
We can use PCA to reduce three features into two principal components.

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

data = {
    "Math": [80, 85, 90, 75, 95],
    "Science": [78, 88, 92, 72, 96],
    "English": [70, 75, 85, 68, 90]
}

df = pd.DataFrame(data)

# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)

# Apply PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

pca_df = pd.DataFrame(
    X_pca,
    columns=["PC1", "PC2"]
)

print(pca_df)

print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)
```

# Interpretation
PCA transforms the original features into new features called Principal Components.

PC1 captures the largest amount of variation, while PC2 captures the next largest amount of variation.

The explained_variance_ratio_ tells us how much of the original data variation is represented by each principal component.

# Business Example
A company may have hundreds of customer-related features. PCA can reduce these features to a smaller number of components while preserving much of the important variation.

# AI/ML Use Case
PCA is used for dimensionality reduction, data visualization, noise reduction, and improving computational efficiency when datasets contain many correlated features.

# 13. Recommendation Systems
Purpose
A recommendation system suggests products, movies, songs, or other items to users based on their preferences and behavior.

Statistics helps recommendation systems identify patterns, similarities, ratings, and user preferences.

How Statistics is Used
Statistics can be used to analyze:

User ratings

Purchase history

Clicks and views

Item popularity

Similarity between users or items

Average ratings and preferences

Two common approaches are:

Collaborative Filtering: recommends items based on similarities between users or items.
    
Content-Based Filtering: recommends items based on the characteristics of items a user has interacted with.
# When to Use
Recommendation systems are useful when a platform has information about users and items.

Examples include:

E-commerce
Movie streaming
    
Music platforms
    
Online learning platforms
# Advantages
Provides personalized recommendations.
    
Helps users discover relevant items.
    
Can improve user engagement.
# Limitations
New users or new items may have little available data.
    
Poor-quality data can produce poor recommendations.
    
Very large datasets can require significant computation.
    
# Python Example
A simple recommendation approach can calculate the average rating for each movie and recommend movies with higher average ratings.

```python
import pandas as pd

data = {
    "Movie": [
        "Movie A", "Movie A",
        "Movie B", "Movie B",
        "Movie C", "Movie C"
    ],
    "Rating": [5, 4, 3, 4, 5, 5]
}

df = pd.DataFrame(data)

average_ratings = df.groupby("Movie")["Rating"].mean()

recommendations = average_ratings.sort_values(ascending=False)

print("Average Ratings:")
print(average_ratings)

print("\nRecommended Movies:")
print(recommendations)
```

# Interpretation
The average rating is calculated for each movie.

Movies with higher average ratings can be considered for recommendation in this simple example.

Real recommendation systems use much more information, such as user behavior, item similarity, previous interactions, and personalized preferences.

# Business Example
An e-commerce website can analyze customer purchases and ratings to recommend products that may be relevant to each customer.

# AI/ML Use Case
Recommendation systems are used by e-commerce, streaming, music, and learning platforms to provide personalized content or product recommendations.

Statistics helps analyze user behavior and measure how well recommendations perform.
